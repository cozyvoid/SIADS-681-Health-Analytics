````python
data = dc.Diabetes()

X = data.X
y = data.y

pid_to_indices = {}
for pid in X['patient_id'].unique().tolist():
    X_df = X[X['patient_id'] == pid]
    pid_to_indices[pid] = X_df.index

splits = dc.TrainSplits(X, y, combine=True, pid_to_indices=pid_to_indices)
````
