### 1. The CDC ensemble generated predictions for:

    > Each county in the US

        > Lucas et al. evaluated the CDC’s COVIDhub ensemble using county-level forecasts.

### 2. The baseline CDC model

    > 	Predicts that cases will be exactly what they were the previous week

        > This is a persistence baseline. If a county reported 100 cases last week, the baseline forecasts 100 for the next week.

### 3. The authors list which of the following as potential sources of variance in the data:

    > All four — lags in reporting, incubation time of the virus, availability of tests, and asymptomatic cases

        > Section 5.5 names all four as sources of variation in reported case counts.

### 4. Which of the following steps did the authors take to overcome issues posed by noisy time series data?

    > Computed a 7-day rolling average only
    > Removed instances with a weekly increase of 0

        > The authors smoothed the daily case data with a 7-day rolling average, then excluded training sequences containing a week with zero reported new cases.

### 5. Looking at figure 9, what might another reasonable choice for the temporal lag be?

    > Lag 10

        > Figure 9 and Table 5 support a 10-week lag as an alternative. It had an average test MAE of 86.05, compared with 87.29 for the 9-week lag chosen through cross-validation on training data.
