Obermeyer et al., *“Dissecting racial bias in an algorithm used to manage the health of populations.”*

### 1. The authors audited an algorithm for:

- Predicting the future health costs for a patient

The algorithm’s label/target was future healthcare expenditures (cost), which it used as a proxy for health need.

### 2. We can interpret the model to be well calibrated for costs if:

- For all patients with predicted cost in the range [s1,s2], the true average cost should be in the range [s1,s2]

Calibration concerns whether average observed outcomes correspond to predicted outcomes among patients receiving similar predictions—not whether every individual's prediction is exact.

### 3. What does figure 3A show?

- Considering cost, the model is well calibrated for white patients and is well calibrated for Black patients



Figure 3A shows relatively little racial disparity in cost conditional on risk score. This is the central paradox: the model predicts cost reasonably well while still being biased as a measure of health need.

### 4. The authors found that Black patients incurred different types of costs than white patients.

- True, Black patients were more likely to have costs related to emergency visits and dialysis

Black patients had fewer inpatient surgical/outpatient specialist costs but more emergency and dialysis costs.

### 5. Figure 1A shows that:

- Overall, at a given risk score white patients are more healthy than Black patients



Figure 1A shows that Black patients have more active chronic conditions at the same algorithm risk score. Thus, a Black and white patient receiving the same score are not equally healthy.

### 6. Figure 2 shows that:

- Across health markers, Black patients with the same risk score as white patients experience worse health conditions



Figure 2 examines multiple markers—hypertension, diabetes severity, cholesterol, renal failure, anemia—and consistently shows greater health need among Black patients at comparable scores.

### 7. The authors find that the most likely mechanism which contributed to bias was:

- Differences in the costs the medical system expends on members of the Black population and members of the white population

The key mechanism is that Black patients generate lower healthcare expenditures at the same level of illness/health need. Cost therefore becomes a racially biased proxy for health.

### 8. The authors suggest that to mitigate bias:

- Training data labels are chosen carefully. For example, it is really important to make sure that your label accurately captures the outcome you care about

Their proposed solution focuses on label choice. If the goal is health need, the target should represent health need rather than relying on cost as a proxy.

### 9. The authors claim that:

- Even reasonable labels can introduce bias

This is one of the major conclusions. Cost seemed like a reasonable, predictive proxy for health, yet structural inequalities caused it to generate substantial racial bias.

### 10. Select all that apply.
### The authors show that:

- [x] Providing doctors with patient records and risk scores mitigates the bias of the algorithm
- [x] Sampling patients according to the number of active conditions they have would mitigate the bias of the algorithm

Doctors partially corrected the algorithm's bias using additional patient information, although only modestly. Prioritizing predicted health/active chronic conditions produced substantially greater mitigation. Random sampling did not solve the underlying problem, and simply balancing the training data was not the authors' proposed solution.
