### 6. Our goal is to predict liver cancer for Livermore hospital. We train a model on data from this hospital and our model achieves an F-measure of .45. To improve the F-measure we try to transfer knowledge from Cancermore hospital.

    > Our new F-measure is .40
    > Our new F-measure is .3

### 7. The transfer learning setting described by "A Deep Learning Approach for Blood Glucose Prediction of Type 1 Diabetes" is an example of:

    > Homogenous transfer learning
    > Parameter based transfer learning

        >> The input features remain the same, and pretrained model weights transfer to individual models. <<

### 8. To classify whether a medical record contains evidence of a patient having a stroke we train a deep neural network on data from Livermore hospital. Instead of using the documents themselves we first embed these documents into a feature representation according to a model trained on documents from hospitals across Michigan used for a variety of tasks.

#### This is an example of transfer learning.

    > Yes, because the tasks are different from the source to the target
    > Yes, because the domain is different from the source to the target

### 9. In transfer learning we:

    > Seek to optimize performance for the target task

        >> Transferred knowledge is intended to improve the task you want to solve. <<

### 10. Transfer learning generally leads to more biased algorithms.

    > False

        >> Transfer learning does not inherently make algorithms more biased; its effects depend on the source data and target setting. <<
