### 1. This paper proposes a method for:

> Predicting the diagnostic codes associated with a clinical document

CAML predicts ICD diagnostic codes from clinical text, such as hospital discharge summaries.

### 2. The authors mention the following reasons as contributing to the difficulty of predicting codes from clinical text:

> Clinical text contains a lot of irrelevant information
> There are tens of thousands of codes yet only a few are relevant per document

Clinical notes contain substantial irrelevant information. Additionally, there are tens of thousands of ICD codes, but only a small subset applies to any individual document.

### 3. The authors treat code explanation as:

> An unsupervised learning problem

CAML generates code-specific explanations through attention without requiring separately labeled training examples identifying which words explain each code. The explanation mechanism does not use direct explanation supervision.

### 4. The authors treat ICD prediction as:

> A multi-label classification problem

One clinical document can have multiple diagnoses, meaning multiple ICD codes may be assigned simultaneously.

### 5. The authors reduce the document size in CAML by:

>  None of the above

CAML uses label-wise attention to aggregate information across the document, rather than average, max, or min pooling. Max pooling is used in the baseline CNN, not CAML's primary attention mechanism.

### 6. The attention mechanism in CAML:

> Finds the argmax of a_l

For each code $ l $, the attention vector $a_l$ assigns importance weights to text positions. The position with the largest attention weight is used to identify the most relevant text for the predicted code.

### 7. The CAML paper is an example of:

> Explainable machine learning

CAML produces predictions using a neural network and provides code-specific explanations through attention weights, making it an example of explainable ML.

### 8. Select all examples of multimodal deep learning mentioned in the *Multimodal Deep Learning* paper:

> Cross modality learning
> Fusion learning
> Shared representation learning

Ngiam et al. describe three multimodal learning settings: learning shared representations across modalities, learning when multiple modalities are combined (fusion), and learning to use one modality to predict or reconstruct another (cross modality).

### 9. Select all which are discussed in the *Multimodal Deep Learning* paper:

> The model shown in figure 3b learns meaningful multi-modal relatinships.
> The model shown in figure 2c learns meaningful single-modality relationships.

Figure 3b shows a deeper bimodal autoencoder capable of learning meaningful relationships between modalities. 
Figure 2c illustrates a shallow bimodal RBM that primarily captures single-modality relationships rather than effectively learning cross-modal correlations.

### 10. In *"Interpretable Machine Learning: Fundamental Principles and 10 Grand Challenges"* the authors state:

> A challenge with explainable AI is that debugging is difficult as it is not clear if the model or explainer is wrong

Post-hoc explanations can be misleading or inaccurate. When an explanation appears incorrect, determining whether the underlying model or the explanation method is responsible becomes difficult.

### 11. In *"Interpretable Machine Learning: Fundamental Principles and 10 Grand Challenges"* the authors state that the goal of interpretable ML is to build trust.

> False

The authors argue that the purpose of interpretability is to enable people to understand, evaluate, verify, and troubleshoot model behavior—not simply to create trust. A model should be trusted only when that trust is warranted.

#### Key concepts to remember for exams

- Multi-label classification: One patient note can have multiple ICD codes.
- Unsupervised explanations: CAML's attention mechanism produces explanations without separate explanation labels. This does not mean the ICD prediction task itself is unsupervised.
- Attention versus pooling: CAML uses label-wise attention, allowing different ICD codes to focus on different words.
- Multimodal learning: Combines or relates multiple data modalities, such as audio and video.
- Interpretable versus explainable ML: Interpretability emphasizes understanding a model's actual decision-making process; explainability often involves methods for explaining a complex model.
- Trust versus trustworthiness: A good explanation should support justified evaluation of a model, rather than automatically increase trust.
